```python
# Basic function
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Models and layers
import tensorflow as tf
import tensorflow.keras as kr
from tensorflow.keras.layers import Dense
from tensorflow.keras.models import Model
```

```python
# Dataset processing and loading
from sklearn.datasets import fetch_openml
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
# Load dataset
# X é a matriz de atributos
# y é o vetor de rótulos
X,y=fetch_openml(name='diabetes', version=1, return_X_y=True)
y=LabelEncoder().fit_transform(y)
```

```python
# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

# Normalize features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train = pd.DataFrame(X_train, columns=X.columns)

X_train.info()
print("\n", "-"*100, "\n")

X_train.head()
```

```python
input=kr.Input(shape=(X_train.shape[1],))
output=Dense(1,activation='sigmoid')(input)
nn=Model(inputs=input,outputs=output)

nn.summary()
```

```python
nn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

nn.fit(X_train, y_train, epochs=100, verbose=True, validation_data=(X_test, y_test))
```

```python
np.mean((nn.predict(X_test).flatten()>0.5)==y_test)
```
