```python
import numpy as np
import matplotlib.pyplot as plt

x = 2*np.random.rand(100, 1)
y = 4 + 3*x + np.random.randn(100, 1)

x_b = np.c_[np.ones((100, 1)), x]  # add x0 = 1 em cada instância
theta_best = np.linalg.inv(x_b.T.dot(x_b)).dot(x_b.T).dot(y)

print(theta_best)
```

A função que usamos é y = 4 + 3x_1 + (ruído_gaussiano). O ruído tourno impossível recuperar os parâmetros exatos da função original. Agora podemos fazer algumas previsões usando os parâmetros encontrados.

```python
x_new = np.array([[0], [2]])
x_new_b = np.c_[np.ones((2, 1)), x_new] # add x0 = 1 em cada instância
y_predict = x_new_b.dot(theta_best)
print(y_predict)
print("\n---------------------------------------------------\n")

plt.plot(x_new, y_predict, "r-")
plt.plot(x, y, "b.")
plt.axis([0, 2, 0, 15])
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

lin_reg = LinearRegression()
lin_reg.fit(x, y)
lin_reg.intercept_, lin_reg.coef_
print(lin_reg.intercept_, lin_reg.coef_)
print("\n---------------------------------------------------\n")
lin_reg.predict(x_new)
print(lin_reg.predict(x_new))
```
