```python
import matplotlib.pyplot as plt
import numpy as np

def display_color(R, G, B):
    fig, ax = plt.subplots(figsize=(2, 2))
    color_block = np.ones((100, 100, 3), dtype=np.uint8)
    color_block[:, :, 0] = R  # Red
    color_block[:, :, 1] = G  # Green
    color_block[:, :, 2] = B  # Blue

    ax.imshow(color_block)
    ax.axis('off')  # Desativa os eixos
    plt.show()
```

```python
display_color(100, 100, 100)  # Vermelho
```
